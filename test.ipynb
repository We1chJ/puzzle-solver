```python
import matplotlib.pyplot as plt
from matching import display_images
from matching import template_matching

boxed_images = ["boxed_images/00.jpg", "boxed_images/01.jpg", "boxed_images/02.jpg", "boxed_images/10.jpg", "boxed_images/11.jpg", "boxed_images/12.jpg"]
pieces = ["test/piece_1.png", "test/piece_2.png", "test/piece_3.png", "test/piece_4.png", "test/piece_5.png", "test/piece_6.png"]

original_image, original_boxes, pieces, box_coordinates = display_images(
    "test/original.jpeg",
    boxed_images,
    pieces
)

solved_puzzle, matches = template_matching(
    original_image,
    original_boxes,
    pieces,
    box_coordinates
)

plt.figure(figsize=(10, 8))
plt.imshow(solved_puzzle)
plt.title("Solved Puzzle")
plt.axis("off")
plt.show()
```

```python
import cv2
import testing_matching

# Same test inputs as above, matched with the multi-scale masked matching
# from testing_matching.py instead of matching.template_matching.
# (`pieces` above now holds the loaded arrays, so the paths are listed again.)
piece_paths = ["test/piece_1.png", "test/piece_2.png", "test/piece_3.png", "test/piece_4.png", "test/piece_5.png", "test/piece_6.png"]
original_bgr = cv2.imread("test/original.jpeg")

placements, solved = testing_matching.Solve(original_bgr, piece_paths, rows=2, cols=3)

for path, (row, col) in placements.items():
    print(f"{path}: row {row}, col {col}")

plt.figure(figsize=(10, 8))
plt.imshow(cv2.cvtColor(solved, cv2.COLOR_BGR2RGB))
plt.title("Solved Puzzle (testing_matching)")
plt.axis("off")
plt.show()
```
